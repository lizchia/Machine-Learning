# OpenAI compabile API

## load API key

```python
import os
from dotenv import load_dotenv

# 加載 .env 檔案
# 在同一個資料夾內編輯一個 .env 文字檔案，可以輸入類似： TMUNLPKEY=XXX
# 記得將 XXX 改成我發給你的 api key (含 sk- 開頭的一連串亂碼)
load_dotenv()


llama8b = {'model_name':'llama8b', 'url':'http://192.168.150.73:4000/v1', 'api_key':os.getenv("TMUNLPKEY")}
llama70b = {'model_name':'llama70b', 'url':'http://192.168.150.73:4000/v1', 'api_key':os.getenv("TMUNLPKEY")}
gptoss20b = {'model_name':'gpt-oss-20b', 'url':'http://192.168.150.73:4000/v1', 'api_key':os.getenv("TMUNLPKEY")}
```

## LLama 8B

```python
from openai import OpenAI


client = OpenAI(
    base_url=llama8b['url'],
    api_key=llama8b['api_key'],
)

completion = client.chat.completions.create(
    model=llama8b['model_name'],
    messages=[
        {"role": "system", 
         "content": "You are a helpful assistant. 務必全程使用正體中文或英文，絕對不能使用簡體中文。"},
        {"role": "user", 
         "content": "請介紹 RAG 是什麼?"}
    ]
)

print(completion.choices[0].message.content.strip())
```

```python
completion.choices[0].message
```

### GPT OSS 20B (Reasoning Model)

```python
from openai import OpenAI


client = OpenAI(
    base_url=gptoss20b['url'],
    api_key=gptoss20b['api_key'],
)

# 使用 GPT OSS Reasoning function

completion = client.chat.completions.create(
    model=gptoss20b['model_name'],
    messages=[
        {"role": "system", 
         "content": "Reasoning: high\n\nYou are a helpful assistant. 務必全程使用正體中文或英文，絕對不能使用簡體中文。"},
        {"role": "user", 
         "content": "請介紹 RAG 是什麼?"}
    ]
)

print(completion.choices[0].message.content.strip())
```

```python
completion.choices[0].message.reasoning_content
```
