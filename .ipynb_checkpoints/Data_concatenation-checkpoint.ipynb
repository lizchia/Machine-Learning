```python
import pandas as pd
import numpy as np
import csv
import sys
import gzip
import os

# 設定常數：篩選的時間窗口
PATH_ICUSTAYS = 'icustays.csv'
PATH_ADMISSIONS = 'admissions.csv'
PATH_PATIENTS = 'patients.csv'
PATH_PROCEDURES = 'procedureevents.csv' # 或 ventilation.csv
PATH_RADIOLOGY = 'radiology.csv.gz'     # 支援 .csv 或 .csv.gz

# 參數設定
CXR_WINDOW_START = -6   # 插管前 6 小時 (容許急診插管的時間差)
CXR_WINDOW_END = 48     # 插管後 48 小時
MIN_LOS_DAYS = 1        # 排除住院 < 1 天
MIN_SURVIVAL_HOURS = 48 # 排除 48 小時內死亡

# 增加 CSV 欄位讀取上限 (避免報告太長導致錯誤)
csv.field_size_limit(sys.maxsize)

# ==========================================
# 第一步：建立基礎世代 (Base Cohort) & 人口學篩選
# ==========================================
def step1_build_cohort(df_icustays, df_admissions, df_patients):
    print("--- [Step 1] 建立基礎世代 (Expanded Cohort) ---")
    
    # 欄位標準化
    for df in [df_icustays, df_admissions, df_patients]:
        df.columns = df.columns.str.lower().str.strip()

    # 1. 準備資料
    # 優先使用 patients 的 dod (死亡日期)，若無則用 admissions 的 deathtime
    df_pat = df_patients[['subject_id', 'anchor_age', 'gender', 'dod']].copy()
    
    # 串接
    df = df_icustays.merge(df_admissions[['subject_id', 'hadm_id', 'hospital_expire_flag']], 
                           on=['subject_id', 'hadm_id'], how='left')
    df = df.merge(df_pat, on='subject_id', how='left')
    
    # 時間轉換
    for col in ['intime', 'outtime', 'dod']:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors='coerce')

    # 處理死亡時間
    df = df.rename(columns={'dod': 'deathtime'})

    # 2. 篩選成年人
    df = df[df['anchor_age'] >= 18]
    
    # 3. 篩選病房 (擴大納入 MICU & SICU，排除心臟/神經)
    # 邏輯：要是內/外科，但不能是純心臟/神經科
    target_pattern = 'MICU|Medical|SICU|Surgical'
    exclude_pattern = 'Cardiac|Coronary|Neuro|Trauma|CSRU|CVICU'
    
    mask_target = df['first_careunit'].str.contains(target_pattern, case=False, na=False)
    mask_exclude = df['first_careunit'].str.contains(exclude_pattern, case=False, na=False)
    
    df = df[mask_target & (~mask_exclude)]
    
    print(f"篩選後病患基數: {len(df)}")
    return df
```

```python
# ==========================================
# 第二步：串接插管資料 (Ventilation Event)
# ==========================================
def step2_merge_ventilation(df_cohort, df_procedures):
    print("\n--- [Step 2] 串接插管時間 ---")
    df_procedures.columns = df_procedures.columns.str.lower().str.strip()
    
    # 篩選插管 (Itemid 224385)
    intubations = df_procedures[df_procedures['itemid'] == 224385].copy()
    intubations['starttime'] = pd.to_datetime(intubations['starttime'], errors='coerce')
    
    # 合併並取「第一次」插管
    df_merged = df_cohort.merge(intubations[['stay_id', 'starttime']], on='stay_id', how='inner')
    df_merged = df_merged.sort_values('starttime')
    df_merged = df_merged.drop_duplicates(subset=['stay_id'], keep='first')
    df_merged = df_merged.rename(columns={'starttime': 'intubation_time'})
    
    print(f"插管病患人數: {len(df_merged)}")
    return df_merged
```

```python
def step2_5_load_radiology_rescue(path_to_csv, target_subject_ids):
    print("\n--- [Step 2.5] 讀取 X 光報告 (GZIP Rescue Mode) ---")
    print(f"目標搜尋名單: {len(target_subject_ids)} 人")
    
    target_subjects_set = set(target_subject_ids)
    collected_rows = []
    header = []
    total_lines = 0
    saved_count = 0
    
    # 判斷是否為壓縮檔
    if path_to_csv.endswith('.gz'):
        open_func = gzip.open
        mode = 'rt' # text mode
    else:
        open_func = open
        mode = 'r'

    try:
        with open_func(path_to_csv, mode, encoding='utf-8', errors='replace') as f:
            reader = csv.reader(f)
            
            try:
                header = next(reader)
                header = [h.lower().strip() for h in header]
                
                # 定位欄位
                if 'subject_id' not in header: 
                    print("❌ 錯誤：找不到 subject_id")
                    return pd.DataFrame()
                
                idx_subj = header.index('subject_id')
                idx_text = header.index('text') if 'text' in header else -1
                
            except StopIteration:
                return pd.DataFrame()

            # 逐行掃描
            for row in reader:
                total_lines += 1
                if total_lines % 100000 == 0:
                    print(f"已掃描 {total_lines} 行... (捕獲 {saved_count} 筆)")
                
                try:
                    # 檢查 ID
                    if len(row) <= idx_subj: continue
                    curr_id_str = row[idx_subj]
                    if not curr_id_str.isdigit(): continue
                    
                    if int(curr_id_str) in target_subjects_set:
                        # 檢查關鍵字 (Chest)
                        if idx_text != -1 and len(row) > idx_text:
                            text_content = row[idx_text].lower()
                            if 'chest' in text_content or 'portable' in text_content:
                                collected_rows.append(row)
                                saved_count += 1
                        else:
                            # 若無 text 欄位或欄位不全，只要 ID 對也先留著
                            collected_rows.append(row)
                            saved_count += 1
                            
                except Exception:
                    continue # 跳過壞掉的行

    except Exception as e:
        print(f"⚠️ 讀取中斷 (預期中): {e}")
    
    print(f"掃描結束，共捕獲: {saved_count} 筆報告")
    
    if collected_rows:
        return pd.DataFrame(collected_rows, columns=header)
    return pd.DataFrame()
```

```python
# ==========================================
# 第三步：串接 X 光報告 (CXR Matching)
# ==========================================
def step3_match_cxr(df_cohort, df_radiology):
    print("\n--- [Step 3] 時間配對 (-6h ~ +48h) ---")
    
    # 1. 型別統一
    df_radiology.columns = df_radiology.columns.str.lower().str.strip()
    
    df_cohort['subject_id'] = df_cohort['subject_id'].astype('int64')
    df_radiology['subject_id'] = pd.to_numeric(df_radiology['subject_id'], errors='coerce')
    df_radiology = df_radiology.dropna(subset=['subject_id'])
    df_radiology['subject_id'] = df_radiology['subject_id'].astype('int64')
    
    # 2. 時間轉換
    df_radiology['charttime'] = pd.to_datetime(df_radiology['charttime'], errors='coerce')
    df_radiology = df_radiology.dropna(subset=['charttime'])
    
    # 3. 合併
    id_col = 'note_id' if 'note_id' in df_radiology.columns else 'report_id'
    cols = ['subject_id', 'charttime', 'text', id_col]
    cols = [c for c in cols if c in df_radiology.columns] # 防呆
    
    df_merged = df_cohort.merge(df_radiology[cols], on='subject_id', how='left')
    
    # 4. 時間窗口篩選
    df_merged['intubation_time'] = pd.to_datetime(df_merged['intubation_time'])
    
    # 邏輯： X光時間必須在 [插管-6h, 插管+48h] 之間
    t_start = df_merged['intubation_time'] + pd.Timedelta(hours=CXR_WINDOW_START)
    t_end = df_merged['intubation_time'] + pd.Timedelta(hours=CXR_WINDOW_END)
    
    mask_valid = (df_merged['charttime'] >= t_start) & (df_merged['charttime'] <= t_end)
    valid_cxr = df_merged[mask_valid].copy()
    
    # 5. 取最接近插管時間的一筆 (Abs Diff)
    valid_cxr['time_diff'] = (valid_cxr['charttime'] - valid_cxr['intubation_time']).abs()
    valid_cxr = valid_cxr.sort_values(['subject_id', 'time_diff'])
    valid_cxr = valid_cxr.drop_duplicates(subset=['stay_id'], keep='first')
    
    # 6. 確保有文字
    valid_cxr = valid_cxr[valid_cxr['text'].notna()]
    if id_col in valid_cxr.columns:
        valid_cxr = valid_cxr.rename(columns={id_col: 'report_id'})
        
    print(f"配對後報告數: {len(valid_cxr)}")
    return valid_cxr
```

```python
# ==========================================
# 第四步：執行嚴格排除標準 (Exclusion Criteria)
# ==========================================
def step4_apply_exclusions(df):
    print("\n--- [Step 4] 臨床排除標準 ---")
    n_start = len(df)
    
    # 1. 排除早期死亡 (<48h)
    time_to_death = df['deathtime'] - df['intubation_time']
    # 保留: (還活著 NaT) OR (活超過 48h)
    mask_alive = (df['deathtime'].isna()) | (time_to_death >= pd.Timedelta(hours=MIN_SURVIVAL_HOURS))
    df = df[mask_alive]
    print(f"排除早期死亡後: {len(df)}")
    
    # 2. 排除極短住院 (<1 day)
    df['icu_los_days'] = (df['outtime'] - df['intime']).dt.total_seconds() / 86400
    df = df[df['icu_los_days'] >= MIN_LOS_DAYS]
    print(f"排除短暫住院後: {len(df)}")
    
    # 3. 排除氣切 (Tracheostomy)
    mask_trach = df['text'].str.lower().str.contains('tracheostomy|trach tube', na=False)
    df = df[~mask_trach]
    print(f"排除氣切後: {len(df)}")
    
    print(f"=== 最終剩餘: {len(df)} 筆 ===")
    return df
```

```python
# ==========================================
# 主執行區塊
# ==========================================
if __name__ == "__main__":
    print(">>> 開始執行 MIMIC-IV ICU LOS 資料串接 Pipeline <<<")
    
    # 1. 讀取 CSV (這裡用 pandas 讀小檔)
    print("讀取基礎表格...")
    df_icustays = pd.read_csv(PATH_ICUSTAYS)
    df_admissions = pd.read_csv(PATH_ADMISSIONS)
    df_patients = pd.read_csv(PATH_PATIENTS)
    df_procedures = pd.read_csv(PATH_PROCEDURES) # 確保這裡有 intubation 資料
    
    # 2. 執行 Step 1 & 2
    df_cohort = step1_build_cohort(df_icustays, df_admissions, df_patients)
    df_cohort = step2_merge_ventilation(df_cohort, df_procedures)
    
    # 3. 取得目標 ID 並執行 Step 2.5 (讀取大檔)
    target_ids = df_cohort['subject_id'].unique()
    
    # 讀取 Radiology (請確認 PATH_RADIOLOGY 指向完整檔案)
    if os.path.exists(PATH_RADIOLOGY):
        df_radiology_full = step2_5_load_radiology_rescue(PATH_RADIOLOGY, target_ids)
        
        if not df_radiology_full.empty:
            # 4. 執行 Step 3 & 4
            df_final = step3_match_cxr(df_cohort, df_radiology_full)
            df_final = step4_apply_exclusions(df_final)
            
            # 5. 存檔
            #output_file = 'final_cohort_data.csv'
            #df_final.to_csv(output_file, index=False)
            #print(f"\n✅ 成功！檔案已儲存至: {output_file}")
            print(f"總資料筆數: {len(df_final)}")
        else:
            print("❌ 錯誤：Radiology 檔案中找不到任何匹配資料。")
    else:
        print(f"❌ 找不到 Radiology 檔案: {PATH_RADIOLOGY}")
```
